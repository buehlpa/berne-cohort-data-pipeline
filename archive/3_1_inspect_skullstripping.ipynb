```python
import matplotlib.pyplot as plt
from ipywidgets import IntSlider, interactive, VBox
from IPython.display import display, clear_output
import os
import numpy as np
os.environ["CUDA_VISIBLE_DEVICES"] = "1"
import nibabel as nib
import plotly.graph_objects as go
from skimage import measure



from concurrent.futures import ThreadPoolExecutor, as_completed
```

```python
# select a file to inspect , 3d render

nifti_file = "/home/bule/berner_daten_analyse/data/nifti_third_try/processed/442968/norm_bet_dir/dwi_bet_normalized.nii.gz"

img = nib.load(nifti_file)
data = img.get_fdata()
```

```python

verts, faces, _, _ = measure.marching_cubes(data, level=np.percentile(data, 75))
fig = go.Figure(
    data=[
        go.Mesh3d(
            x=verts[:, 0], y=verts[:, 1], z=verts[:, 2],
            i=faces[:, 0], j=faces[:, 1], k=faces[:, 2],
            opacity=1.0,  # Transparency
            color='gray'
        )
    ]
)

fig.update_layout(
    scene=dict(
        xaxis=dict(visible=False),
        yaxis=dict(visible=False),
        zaxis=dict(visible=False),
        aspectmode="data"
    )
)
fig.show()
```
