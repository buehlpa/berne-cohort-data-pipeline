this file takes the selection json  created by the app_manual_preselection.py and saves the overlapping sequences in a dataframe

```python
import os
os.environ["CUDA_VISIBLE_DEVICES"] = "1"

import pandas as pd
from datetime import datetime
import matplotlib.image as mpimg
import matplotlib.pyplot as plt
import json

from utils.selection import *


# Source of RAW DICOMs
BERN_PATH='/nasdata/Berner_Daten/Other_Side_Bilddaten' 
# Project data path
DATA_PATH='/home/bule/berner_daten_analyse/data'

APP_DATA_PATH = '/home/bule/berner_daten_analyse/apps/app_manual_preselection/selection_1quartal2025'
```

## Selection by clinicians with domain knowledge 

- rater 1

```python
with open(os.path.join(APP_DATA_PATH,'rater1_sequences.json')) as json_file:
    selection_rater1_dict = json.load(json_file)

rows = []
for patid, selection_dict in selection_rater1_dict.items():
    for modality, sequences in selection_dict.items():
        for sequence in sequences:
            study_id, sequence_id = sequence.split('/')
            rows.append({
                'Patient_ID': patid,
                'Study_ID': study_id,
                'Sequence_ID': sequence_id,
                'Modality': modality,
                'Labeller': 'rater1'  
            })
selection_rater1_df = pd.DataFrame(rows).drop_duplicates()
selection_rater1_df['Sequence_path']=selection_rater1_df.apply(lambda x: os.path.join(BERN_PATH,x['Patient_ID'],x['Study_ID'],x['Sequence_ID']),axis=1)
selection_rater1_df['Date'] = selection_rater1_df['Study_ID'].str[:8].apply(lambda x: datetime.strptime(x, '%Y%m%d').strftime('%Y-%m-%d'))

# Filter to keep only dwi and t1 modalities
df_filtered = selection_rater1_df[selection_rater1_df["Modality"].isin(["dwi", "t1"])]

first_dwi = df_filtered[df_filtered["Modality"] == "dwi"].sort_values(["Patient_ID", "Date"]).groupby("Patient_ID").first().reset_index()
first_t1 = df_filtered[df_filtered["Modality"] == "t1"].sort_values(["Patient_ID", "Date"]).groupby("Patient_ID").first().reset_index()

# only get thoes with exactly 2 mods
paired_patients_rater1_ids = pd.merge(first_dwi, first_t1, on="Patient_ID", suffixes=("_dwi", "_t1"))['Patient_ID']
paired_patients_rater1=pd.concat([first_dwi,first_t1])
paired_patients_rater1=paired_patients_rater1[paired_patients_rater1["Patient_ID"].isin(paired_patients_rater1_ids)]
unpaired_patients_rater1=selection_rater1_df[~selection_rater1_df["Patient_ID"].isin(paired_patients_rater1_ids)]
paired_patients_rater1.sort_values("Patient_ID", inplace=True)
paired_patients_rater1.info()
```

- save mergefile

```python
dwi_t1_df_merged = []  # 
for _, row in selection_merged_df.iterrows():
    for dcm_ending in os.listdir(row['Sequence_path']):        
        dwi_t1_df_merged.append({
            'Patient_ID': row['Patient_ID'],
            'Study_ID': row['Study_ID'],
            'Sequence_ID': row['Sequence_ID'],
            'File_Name': dcm_ending,
            'Full_Path': os.path.join(row['Sequence_path'], dcm_ending),
            'Modality': row['Modality'].upper()
        })

dwi_t1_df_merged = pd.DataFrame(dwi_t1_df_merged)
dwi_t1_df_merged['Date'] = dwi_t1_df_merged['Study_ID'].str[:8].apply(lambda x: datetime.strptime(x, '%Y%m%d').strftime('%Y-%m-%d'))
dwi_t1_df_merged['Sequence_ID_split'] = dwi_t1_df_merged['Sequence_ID'].apply(lambda x: split_at_anynumber_seq(x))

dwi_t1_df_merged["File_Number"] = dwi_t1_df_merged["File_Name"].str.extract(r'(\d+)').astype(int)
dwi_t1_df_merged.sort_values(["Patient_ID", "Modality", "File_Number"], inplace=True)
dwi_t1_df_merged.drop(columns=["File_Number"], inplace=True)
dwi_t1_df_merged.reset_index()
dwi_t1_df_merged.to_csv(os.path.join(DATA_PATH, 'dwi_t1_df_merged.csv'), index=False)
```
